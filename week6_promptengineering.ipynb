```python
# 하드코딩 말고 입력을 받아보자

prompt = input("Prompt : ") # "안녕. 오늘의 날씨를 알려줘"

response = client.responses.create(
    model = "gpt-4o-mini",
    temperature = 1.0,
    input = prompt
)

print(response.output_text)
```

```python
order = "너는 이제부터 들어오는 입력에 대해 동아리 khuda와 관련되었다면 응답을 생성하고 그렇지 않으면, ‘해당 질문에 대해서는 대답할 수 없습니다.’라고 해."
prompt = input("Prompt : ") # "안녕. 오늘의 날씨를 알려줘"

response = client.responses.create(
    model = "gpt-4o-mini",
    temperature = 1.0,
    input = order + prompt
)

print(response.output_text)
```

```python
# order의 우선순위가 항상 더 높다

order = "너는 이제부터 들어오는 입력에 대해 동아리 khuda와 관련되었다면 응답을 생성하고 그렇지 않으면, ‘해당 질문에 대해서는 대답할 수 없습니다.’라고 해."
prompt = input("Prompt : ") # "안녕. 오늘의 날씨를 알려줘"

response = client.responses.create(
    model = "gpt-4o-mini",
    temperature = 1.0,

    instructions = order,
    input = prompt
)

print(response.output_text)
```
